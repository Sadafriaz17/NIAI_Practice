## Task 53: Machine Learning: Decision Tree
Train a decision tree and draw it.

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.3, random_state=0)

tree = DecisionTreeClassifier(criterion="gini", max_depth=3, random_state=0)
tree.fit(X_train, y_train)
pred = tree.predict(X_test)
print("Accuracy:", accuracy_score(y_test, pred))
print(classification_report(y_test, pred, target_names=iris.target_names))

plt.figure(figsize=(12, 6))
plot_tree(tree, feature_names=iris.feature_names, class_names=iris.target_names, filled=True)
plt.show()

# Which features matter most
for name, score in zip(iris.feature_names, tree.feature_importances_):
    print(f"{name}: {score:.3f}")
```

## Task 54: Machine Learning: Support Vector Machine
Compare a linear kernel with an RBF kernel.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

X, y = make_moons(n_samples=300, noise=0.25, random_state=0)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
for ax, kernel in zip(axes, ["linear", "rbf"]):
    svm = SVC(kernel=kernel, C=1.0)
    svm.fit(X_train, y_train)
    acc = accuracy_score(y_test, svm.predict(X_test))
    print(f"{kernel} kernel accuracy: {acc:.3f}")

    xx, yy = np.meshgrid(np.linspace(X[:, 0].min() - 0.5, X[:, 0].max() + 0.5, 200),
                         np.linspace(X[:, 1].min() - 0.5, X[:, 1].max() + 0.5, 200))
    zz = svm.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
    ax.contourf(xx, yy, zz, alpha=0.3)
    ax.scatter(X_test[:, 0], X_test[:, 1], c=y_test, edgecolor="k")
    ax.set_title(f"SVM {kernel} kernel, accuracy {acc:.2f}")
plt.show()
```

## Task 55: Machine Learning: Time Series Analysis
Work with dates, resample, smooth and find trends.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Make daily sales data with a trend and a weekly pattern
np.random.seed(3)
dates = pd.date_range("2025-01-01", periods=365, freq="D")
trend = np.linspace(100, 200, 365)
weekly = 15 * np.sin(2 * np.pi * np.arange(365) / 7)
noise = np.random.normal(0, 8, 365)
sales = pd.Series(trend + weekly + noise, index=dates, name="sales")
print(sales.head())

# Resample to monthly average
monthly = sales.resample("ME").mean()
print(monthly.head())

# Rolling average, shift and difference
rolling = sales.rolling(window=7).mean()
change = sales.diff()
lagged = sales.shift(1)
print("Largest daily jump:", change.max())

# Plot
fig, axes = plt.subplots(2, 1, figsize=(11, 7))
sales.plot(ax=axes[0], alpha=0.5, label="Daily")
rolling.plot(ax=axes[0], color="red", label="7 day average")
axes[0].legend()
axes[0].set_title("Daily sales")
monthly.plot(ax=axes[1], marker="o")
axes[1].set_title("Monthly average sales")
plt.tight_layout()
plt.show()

# Simple forecast: linear trend
from sklearn.linear_model import LinearRegression
X = np.arange(len(sales)).reshape(-1, 1)
model = LinearRegression().fit(X, sales.values)
future = np.arange(len(sales), len(sales) + 30).reshape(-1, 1)
print("Forecast for the next 5 days:", model.predict(future)[:5].round(1))
```
