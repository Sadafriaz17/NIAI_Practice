## Task 8: Python: Area of a Triangle
Use the semi-perimeter formula.

```python
# Python Program to find the area of triangle

a = 5
b = 6
c = 7

# Uncomment below to take inputs from the user
# a = float(input('Enter first side: '))
# b = float(input('Enter second side: '))
# c = float(input('Enter third side: '))

# calculate the semi-perimeter
s = (a + b + c) / 2

# calculate the area
area = (s * (s - a) * (s - b) * (s - c)) ** 0.5
print('The area of the triangle is %0.2f' % area)
```

## Task 9: Python: Quadratic Equation
Solve ax**2 + bx + c = 0.

```python
# Solve the quadratic equation ax**2 + bx + c = 0

# import complex math module
import cmath

a = 1
b = 5
c = 6

# calculate the discriminant
d = (b ** 2) - (4 * a * c)

# find two solutions
sol1 = (-b - cmath.sqrt(d)) / (2 * a)
sol2 = (-b + cmath.sqrt(d)) / (2 * a)

print('The solution are {0} and {1}'.format(sol1, sol2))
```

## Task 15: Python: For Loops
Practice for loops.

```python
# 1. Loop over a range
for i in range(1, 6):
    print("Number:", i)

# 2. Sum of numbers from 1 to 10
total = 0
for n in range(1, 11):
    total += n
print("Sum 1 to 10 =", total)

# 3. Multiplication table of 5
for n in range(1, 11):
    print(f"5 x {n} = {5 * n}")

# 4. Loop over a string
for ch in "Python":
    print(ch, end=" ")
print()

# 5. Range with a step
for n in range(10, 0, -2):
    print(n, end=" ")
print()

# 6. Loop over a list
fruits = ["apple", "banana", "mango"]
for f in fruits:
    print(f.upper())
```

## Task 16: Python: While Loops
Practice while loops.

```python
# 1. Count down
count = 5
while count > 0:
    print(count)
    count -= 1
print("Go!")

# 2. Sum of digits of a number
number = 4829
total = 0
while number > 0:
    total += number % 10
    number //= 10
print("Sum of digits =", total)

# 3. Factorial
n = 6
fact = 1
i = 1
while i <= n:
    fact *= i
    i += 1
print(f"{n}! = {fact}")

# 4. While with else
x = 0
while x < 3:
    print("x is", x)
    x += 1
else:
    print("loop finished without break")
```

## Task 17: Python: Break Statement
Stop a loop early with break.

```python
# 1. Stop when a letter is found
for ch in "programming":
    if ch == "a":
        print("Found 'a', stopping")
        break
    print(ch)

# 2. First number divisible by 7
numbers = [12, 25, 31, 42, 56, 63]
for n in numbers:
    if n % 7 == 0:
        print("First multiple of 7 is", n)
        break

# 3. Stop a while loop when the total passes 20
total = 0
n = 1
while True:
    total += n
    if total > 20:
        print("Total passed 20 at n =", n)
        break
    n += 1
```

## Task 18: Python: Continue Statement
Skip one round of a loop with continue.

```python
# 1. Print only odd numbers
for n in range(1, 11):
    if n % 2 == 0:
        continue
    print(n, end=" ")
print()

# 2. Skip vowels in a word
for ch in "continue":
    if ch in "aeiou":
        continue
    print(ch, end="")
print()

# 3. Skip negative numbers
values = [4, -1, 7, -3, 9]
for v in values:
    if v < 0:
        continue
    print("Positive value:", v)
```

## Task 19: Python: Looping Techniques
enumerate, zip, reversed, sorted, items and nested loops.

```python
names = ["Ali", "Sara", "Hina"]
marks = [88, 92, 79]

# enumerate
for index, name in enumerate(names, start=1):
    print(index, name)

# zip
for name, mark in zip(names, marks):
    print(name, "scored", mark)

# reversed and sorted
for n in reversed(names):
    print(n, end=" ")
print()
for m in sorted(marks):
    print(m, end=" ")
print()

# dictionary items
student = {"name": "Sadaf", "semester": 8}
for key, value in student.items():
    print(key, "->", value)

# nested loop: simple pattern
for i in range(1, 5):
    for j in range(i):
        print("*", end="")
    print()
```

## Task 21: Python: List Data Type
Create and change lists.

```python
nums = [5, 3, 8, 1, 9]
print("List:", nums, "Length:", len(nums))

# Indexing and slicing
print(nums[0], nums[-1], nums[1:4], nums[::-1])

# Add items
nums.append(7)
nums.insert(2, 100)
nums.extend([4, 6])
print("After adding:", nums)

# Remove items
nums.remove(100)
last = nums.pop()
print("After removing:", nums, "popped:", last)

# Sorting
nums.sort()
print("Sorted:", nums)
nums.sort(reverse=True)
print("Reverse sorted:", nums)

# Useful methods
print("Count of 8:", nums.count(8), "Index of 8:", nums.index(8))

# List comprehension
squares = [n ** 2 for n in range(1, 6)]
print("Squares:", squares)
```

## Task 22: Python: Tuple Data Type
Tuples cannot be changed after they are made.

```python
t = (10, 20, 30, 20, 40)
print("Tuple:", t)
print("First:", t[0], "Last:", t[-1], "Slice:", t[1:3])
print("Count of 20:", t.count(20), "Index of 30:", t.index(30))

# Unpacking
a, b, *rest = t
print(a, b, rest)

# Tuples are immutable
try:
    t[0] = 99
except TypeError as e:
    print("Error:", e)

# Change by making a new tuple
t2 = t + (50,)
print("New tuple:", t2)

# Single item tuple needs a comma
single = (5,)
print(type(single))
```

## Task 24: Python: Set Data Type
Sets keep only unique items.

```python
a = {1, 2, 3, 4, 4, 5}
b = {4, 5, 6, 7}
print("a =", a, "b =", b)

a.add(10)
a.discard(1)
print("After add and discard:", a)

print("Union:", a | b)
print("Intersection:", a & b)
print("Difference:", a - b)
print("Symmetric difference:", a ^ b)

print("Is subset:", {4, 5} <= b)
print("Remove duplicates:", set([1, 1, 2, 2, 3]))
```

## Task 25: Python: Dictionary Data Type
Store data as key and value pairs.

```python
student = {"name": "Sadaf", "semester": 8, "city": "Lahore"}
print(student)

# Access
print(student["name"], student.get("cgpa", "not found"))

# Add and update
student["cgpa"] = 3.89
student.update({"semester": 8, "degree": "BS CS"})
print(student)

# Remove
del student["city"]
removed = student.pop("degree")
print("Removed:", removed, "->", student)

# Loops
for k in student:
    print(k, ":", student[k])
print(list(student.keys()), list(student.values()))

# Dictionary comprehension
squares = {n: n ** 2 for n in range(1, 6)}
print(squares)
```
