## Task 56: Neural Network from Scratch
A small neural network built with NumPy only. It learns the XOR problem.

```python
import numpy as np

np.random.seed(1)
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([[0], [1], [1], [0]])

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def sigmoid_grad(a):
    return a * (1 - a)

W1 = np.random.randn(2, 4)
b1 = np.zeros((1, 4))
W2 = np.random.randn(4, 1)
b2 = np.zeros((1, 1))
lr = 0.5

for epoch in range(10000):
    # forward pass
    h = sigmoid(X @ W1 + b1)
    out = sigmoid(h @ W2 + b2)
    loss = np.mean((y - out) ** 2)

    # backward pass
    d_out = (out - y) * sigmoid_grad(out)
    d_h = (d_out @ W2.T) * sigmoid_grad(h)

    W2 -= lr * h.T @ d_out
    b2 -= lr * d_out.sum(axis=0, keepdims=True)
    W1 -= lr * X.T @ d_h
    b1 -= lr * d_h.sum(axis=0, keepdims=True)

    if epoch % 2000 == 0:
        print(f"epoch {epoch}, loss {loss:.4f}")

print("Predictions:", out.round(3).ravel())
print("Rounded:", out.round().ravel())
```

## Task 57: Deep Learning: MLP
A multi layer perceptron on the digits data.

```python
from sklearn.datasets import load_digits
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt

X, y = load_digits(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)
scaler = StandardScaler().fit(X_train)

mlp = MLPClassifier(hidden_layer_sizes=(64, 32), activation="relu", max_iter=300, random_state=0)
mlp.fit(scaler.transform(X_train), y_train)
pred = mlp.predict(scaler.transform(X_test))
print("MLP accuracy:", accuracy_score(y_test, pred))

plt.plot(mlp.loss_curve_)
plt.xlabel("Iteration")
plt.ylabel("Loss")
plt.title("MLP training loss")
plt.show()
```

## Task 58: Deep Learning: Feed Forward Neural Network
A feed forward network built with Keras. It uses dropout to reduce overfitting.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import tensorflow as tf
from tensorflow import keras

X, y = load_breast_cancer(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
scaler = StandardScaler().fit(X_train)
X_train, X_test = scaler.transform(X_train), scaler.transform(X_test)

model = keras.Sequential([
    keras.layers.Input(shape=(X_train.shape[1],)),
    keras.layers.Dense(32, activation="relu"),
    keras.layers.Dropout(0.3),
    keras.layers.Dense(16, activation="relu"),
    keras.layers.Dense(1, activation="sigmoid"),
])
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
model.summary()

history = model.fit(X_train, y_train, epochs=30, batch_size=16, validation_split=0.2, verbose=0)
loss, acc = model.evaluate(X_test, y_test, verbose=0)
print("Test accuracy:", acc)

plt.plot(history.history["loss"], label="train loss")
plt.plot(history.history["val_loss"], label="validation loss")
plt.legend()
plt.title("Feed forward network")
plt.show()
```

## Task 59: Deep Learning: Loss Functions
Common loss functions written in NumPy.

```python
import numpy as np

y_true = np.array([3.0, 5.0, 2.5, 7.0])
y_pred = np.array([2.5, 5.0, 4.0, 8.0])

mse = np.mean((y_true - y_pred) ** 2)
mae = np.mean(np.abs(y_true - y_pred))
rmse = np.sqrt(mse)

def huber(y, p, delta=1.0):
    err = y - p
    small = np.abs(err) <= delta
    return np.mean(np.where(small, 0.5 * err ** 2, delta * (np.abs(err) - 0.5 * delta)))

print("MSE:", mse)
print("RMSE:", rmse)
print("MAE:", mae)
print("Huber:", huber(y_true, y_pred))

# Classification losses
labels = np.array([1, 0, 1, 1])
probs = np.array([0.9, 0.2, 0.6, 0.95])
eps = 1e-12
bce = -np.mean(labels * np.log(probs + eps) + (1 - labels) * np.log(1 - probs + eps))
print("Binary cross entropy:", bce)

# Categorical cross entropy
true_onehot = np.array([[1, 0, 0], [0, 1, 0]])
pred_probs = np.array([[0.7, 0.2, 0.1], [0.1, 0.8, 0.1]])
cce = -np.mean(np.sum(true_onehot * np.log(pred_probs + eps), axis=1))
print("Categorical cross entropy:", cce)

# Hinge loss
y_pm = np.array([1, -1, 1, -1])
scores = np.array([0.8, -0.5, -0.2, 0.3])
print("Hinge loss:", np.mean(np.maximum(0, 1 - y_pm * scores)))
```

## Task 61: Deep Learning: Text Processing
Clean text: lowercase, remove punctuation and stopwords, stem and lemmatize.

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import word_tokenize

for pkg in ["punkt", "punkt_tab", "stopwords", "wordnet"]:
    nltk.download(pkg, quiet=True)

text = "The students are studying machine learning models, and they were running many experiments!"

lower = text.lower()
no_punct = re.sub(r"[^a-z\s]", "", lower)
tokens = word_tokenize(no_punct)
print("Tokens:", tokens)

stop = set(stopwords.words("english"))
filtered = [t for t in tokens if t not in stop]
print("Without stopwords:", filtered)

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
print("Stems:", [stemmer.stem(t) for t in filtered])
print("Lemmas:", [lemmatizer.lemmatize(t, pos="v") for t in filtered])
```

## Task 62: Deep Learning: Text Analysis
Count words, find common pairs and turn text into numbers.

```python
import nltk
from collections import Counter
from nltk import word_tokenize, bigrams
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

for pkg in ["punkt", "punkt_tab", "stopwords"]:
    nltk.download(pkg, quiet=True)

docs = [
    "Machine learning helps computers learn from data.",
    "Deep learning uses neural networks with many layers.",
    "Natural language processing helps computers understand text.",
]

text = " ".join(docs).lower()
tokens = [t for t in word_tokenize(text) if t.isalpha() and t not in stopwords.words("english")]
print("Most common words:", Counter(tokens).most_common(5))
print("Most common word pairs:", Counter(bigrams(tokens)).most_common(3))

# Bag of words
cv = CountVectorizer(stop_words="english")
bow = cv.fit_transform(docs)
print(cv.get_feature_names_out())
print(bow.toarray())

# TF-IDF and document similarity
tfidf = TfidfVectorizer(stop_words="english")
mat = tfidf.fit_transform(docs)
from sklearn.metrics.pairwise import cosine_similarity
print(cosine_similarity(mat).round(2))
```

## Task 63: Deep Learning: Convolutional Neural Network
A small CNN that reads handwritten digits (MNIST).

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras

(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()
x_train = x_train[..., None].astype("float32") / 255.0
x_test = x_test[..., None].astype("float32") / 255.0

model = keras.Sequential([
    keras.layers.Input(shape=(28, 28, 1)),
    keras.layers.Conv2D(16, 3, activation="relu"),
    keras.layers.MaxPooling2D(),
    keras.layers.Conv2D(32, 3, activation="relu"),
    keras.layers.MaxPooling2D(),
    keras.layers.Flatten(),
    keras.layers.Dense(64, activation="relu"),
    keras.layers.Dense(10, activation="softmax"),
])
model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
model.summary()

model.fit(x_train, y_train, epochs=3, batch_size=128, validation_split=0.1, verbose=2)
loss, acc = model.evaluate(x_test, y_test, verbose=0)
print("Test accuracy:", acc)

pred = model.predict(x_test[:6], verbose=0).argmax(axis=1)
fig, axes = plt.subplots(1, 6, figsize=(12, 3))
for ax, img, p in zip(axes, x_test[:6], pred):
    ax.imshow(img.squeeze(), cmap="gray")
    ax.set_title(f"pred {p}")
    ax.axis("off")
plt.show()
```

## Task 64: Deep Learning: CNN on CIFAR-10
Classify small colour images into 10 classes.

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras

(x_train, y_train), (x_test, y_test) = keras.datasets.cifar10.load_data()
x_train, x_test = x_train.astype("float32") / 255.0, x_test.astype("float32") / 255.0
y_train, y_test = y_train.ravel(), y_test.ravel()
classes = ["airplane", "automobile", "bird", "cat", "deer", "dog", "frog", "horse", "ship", "truck"]

model = keras.Sequential([
    keras.layers.Input(shape=(32, 32, 3)),
    keras.layers.Conv2D(32, 3, activation="relu", padding="same"),
    keras.layers.MaxPooling2D(),
    keras.layers.Conv2D(64, 3, activation="relu", padding="same"),
    keras.layers.MaxPooling2D(),
    keras.layers.Conv2D(128, 3, activation="relu", padding="same"),
    keras.layers.MaxPooling2D(),
    keras.layers.Flatten(),
    keras.layers.Dropout(0.4),
    keras.layers.Dense(128, activation="relu"),
    keras.layers.Dense(10, activation="softmax"),
])
model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])

history = model.fit(x_train, y_train, epochs=8, batch_size=128, validation_split=0.1, verbose=2)
loss, acc = model.evaluate(x_test, y_test, verbose=0)
print("Test accuracy:", acc)

plt.plot(history.history["accuracy"], label="train")
plt.plot(history.history["val_accuracy"], label="validation")
plt.legend()
plt.title("CIFAR-10 accuracy")
plt.show()

pred = model.predict(x_test[:8], verbose=0).argmax(axis=1)
fig, axes = plt.subplots(1, 8, figsize=(16, 3))
for ax, img, p, t in zip(axes, x_test[:8], pred, y_test[:8]):
    ax.imshow(img)
    ax.set_title(f"{classes[p]}\n(true: {classes[t]})", fontsize=8)
    ax.axis("off")
plt.show()
```
