## Task 43: Pandas: Series
A Series is one column of data.

```python
import pandas as pd

a = pd.Series([1, 7, 2])
print(a)
print("Second item:", a[1])

# Custom labels
b = pd.Series([10, 20, 30], index=["x", "y", "z"])
print(b)
print(b["y"])

# From a dictionary
calories = {"day1": 420, "day2": 380, "day3": 390}
c = pd.Series(calories)
print(c)
print(pd.Series(calories, index=["day1", "day2"]))
```

## Task 44: Pandas: DataFrames
A DataFrame is a table with rows and columns.

```python
import pandas as pd

data = {"calories": [420, 380, 390], "duration": [50, 40, 45]}
df = pd.DataFrame(data, index=["day1", "day2", "day3"])
print(df)

# Locate rows
print(df.loc["day2"])
print(df.loc[["day1", "day3"]])
print(df.iloc[0])
print(df["calories"])
print(df[df["duration"] > 40])
```

## Task 47: Pandas: Data Cleaning
Fix empty cells, wrong formats, wrong values and duplicates.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Date": ["2020-12-01", "2020-12-02", "20201203", "2020-12-04", "2020-12-04"],
    "Duration": [60, 45, 450, np.nan, np.nan],
    "Calories": [409.1, 479.0, 340.0, np.nan, np.nan],
})
print("Original:\n", df)

# 1. Empty cells
print("\nMissing values:\n", df.isnull().sum())
filled = df.fillna({"Duration": df["Duration"].median(), "Calories": df["Calories"].mean()})

# 2. Wrong format
filled["Date"] = pd.to_datetime(filled["Date"], format="mixed")

# 3. Wrong data
filled.loc[filled["Duration"] > 120, "Duration"] = 120

# 4. Duplicates
print("\nDuplicates:", filled.duplicated().sum())
clean = filled.drop_duplicates().reset_index(drop=True)
print("\nCleaned:\n", clean)

# Drop rows with empty cells instead
print("\nDropna result:\n", df.dropna())
```

## Task 48: Pandas: Correlations
Find how strongly columns move together.

```python
import pandas as pd
import numpy as np

np.random.seed(0)
duration = np.random.randint(30, 120, 50)
df = pd.DataFrame({
    "Duration": duration,
    "Calories": duration * 4 + np.random.randint(-30, 30, 50),
    "Pulse": np.random.randint(80, 150, 50),
})
print(df.corr())

import matplotlib.pyplot as plt
import seaborn as sns
sns.heatmap(df.corr(), annot=True, cmap="Blues")
plt.title("Correlation matrix")
plt.show()
```
