```python
# Install once if needed:  pip install azure-ai-textanalytics
import os
from azure.ai.textanalytics import TextAnalyticsClient
from azure.core.credentials import AzureKeyCredential

# Set these two values after you create an Azure AI Language resource in the Azure portal.
# Never write a real key inside the notebook. Set it as an environment variable instead.
ENDPOINT = os.environ.get("AZURE_LANGUAGE_ENDPOINT", "")
KEY = os.environ.get("AZURE_LANGUAGE_KEY", "")

client = None
if ENDPOINT and KEY:
    client = TextAnalyticsClient(endpoint=ENDPOINT, credential=AzureKeyCredential(KEY))
    print("Azure client ready")
else:
    print("Set AZURE_LANGUAGE_ENDPOINT and AZURE_LANGUAGE_KEY to run the Azure calls.")
```

```python
# Language detection
docs = ["Hello, how are you?", "Bonjour tout le monde", "Mein Name ist Sadaf"]
if client:
    for doc, res in zip(docs, client.detect_language(documents=docs)):
        print(doc, "->", res.primary_language.name, round(res.primary_language.confidence_score, 2))
else:
    print("Skipped: Azure client is not set up yet.")
```
