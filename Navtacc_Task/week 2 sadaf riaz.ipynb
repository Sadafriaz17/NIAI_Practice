## Task 20: User Defined Functions
Functions, default values, scope, lambda, map, filter and nested functions.

```python
# Basic function with a default value
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet("Sadaf"))
print(greet("Sadaf", "Welcome"))

# Return more than one value
def min_max(values):
    return min(values), max(values)

low, high = min_max([4, 9, 1, 7])
print("Min:", low, "Max:", high)

# Variable length arguments
def total(*numbers):
    return sum(numbers)

print("Total:", total(1, 2, 3, 4))

def show(**info):
    for k, v in info.items():
        print(k, "=", v)

show(name="Sadaf", course="AI")
```

```python
# Variable scope
x = "global"

def test_scope():
    x = "local"
    print("Inside function:", x)

test_scope()
print("Outside function:", x)

counter = 0
def increase():
    global counter
    counter += 1

increase(); increase()
print("Counter:", counter)
```

```python
# Lambda, map and filter
square = lambda n: n ** 2
print(square(6))

nums = [1, 2, 3, 4, 5, 6]
print("Squares:", list(map(lambda n: n ** 2, nums)))
print("Even numbers:", list(filter(lambda n: n % 2 == 0, nums)))

# Inner (nested) function
def outer(a):
    def inner(b):
        return a + b
    return inner

add5 = outer(5)
print(add5(10))
```

## Task 26: Exception Handling and File Handling
Handle errors with try, except, else and finally. Read and write a file.

```python
# Exception handling
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Error: cannot divide by zero")
    except TypeError:
        print("Error: please give numbers only")
    else:
        print("Result:", result)
    finally:
        print("Division finished")

divide(10, 2)
divide(10, 0)
divide(10, "a")

try:
    int("abc")
except ValueError as e:
    print("ValueError:", e)
```

```python
# File handling
with open("sample.txt", "w") as f:
    f.write("Line one\n")
    f.write("Line two\n")

with open("sample.txt", "a") as f:
    f.write("Line three\n")

with open("sample.txt", "r") as f:
    content = f.read()
print(content)

with open("sample.txt") as f:
    for line_no, line in enumerate(f, start=1):
        print(line_no, line.strip())

try:
    open("missing_file.txt")
except FileNotFoundError:
    print("File not found, handled safely")
```

## Task 27: User Defined Exception Handling
Make your own exception class.

```python
class NegativeAgeError(Exception):
    def __init__(self, age, message="Age cannot be negative"):
        self.age = age
        self.message = message
        super().__init__(f"{message}: {age}")

def set_age(age):
    if age < 0:
        raise NegativeAgeError(age)
    return f"Age set to {age}"

for value in [22, -5]:
    try:
        print(set_age(value))
    except NegativeAgeError as e:
        print("Caught:", e)
```

## Task 49: Statistics: Mean, Median and Mode
Find the central values of a data set.

```python
import numpy as np
import statistics

speed = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]

print("Mean (numpy):", np.mean(speed))
print("Median (numpy):", np.median(speed))
print("Mode (statistics):", statistics.mode(speed))

# Median with an even number of values
print("Median of even list:", np.median([1, 2, 3, 4]))

# Without libraries
mean_manual = sum(speed) / len(speed)
print("Mean (manual):", round(mean_manual, 2))
```

## Task 50: Statistics: Standard Deviation
Measure how spread out the data is.

```python
import numpy as np

speed = [86, 87, 88, 86, 87, 85, 86]
speed2 = [32, 111, 138, 28, 59, 77, 97]

for data in (speed, speed2):
    print("Data:", data)
    print("  Mean:", round(np.mean(data), 2))
    print("  Variance:", round(np.var(data), 2))
    print("  Standard deviation:", round(np.std(data), 2))

# Manual standard deviation
mean = sum(speed2) / len(speed2)
variance = sum((x - mean) ** 2 for x in speed2) / len(speed2)
print("Manual standard deviation:", round(variance ** 0.5, 2))

# Coefficient of variation
print("Coefficient of variation:", round(np.std(speed2) / np.mean(speed2), 3))
```
