Comments are notes in your code that Python **ignores** when running. They exist purely to help humans (including future-you) understand the code.

- Single-line comment: starts with `#`
- Multi-line comment: usually written as consecutive `#` lines, or a triple-quoted string used as a block explanation

<a id="naming"></a>
# 8. Naming Conventions

Rules for naming variables in Python:
- Must start with a letter or underscore (`_`), not a digit.
- Can contain letters, digits, and underscores.
- Case-sensitive (`age` and `Age` are different variables).
- Cannot be a Python reserved keyword (e.g. `if`, `for`, `class`).

Convention (PEP 8, the Python style guide): use **snake_case** for variable names — lowercase words separated by underscores.

````python
# Good variable names
learning_rate = 0.01
num_students = 30

# Avoid names like these (they still work, but are bad style):
lr = 0.01          # too vague
NumStudents = 30    # not snake_case (this style is reserved for class names)

print(learning_rate, num_students)
````

**Common mistake:** Starting a variable name with a number (`2nd_value = 5`) — this raises a `SyntaxError`. Try it below.

````python
# 2nd_value = 5   # Uncomment to see the SyntaxError, then delete/comment it again
valid_name_2 = 5   # This is fine — digits are allowed, just not as the first character
print(valid_name_2)
````

<a id="type-conversion"></a>
# 11. Type Conversion

Sometimes you need to convert a value from one type to another. Python provides built-in functions for this:

- `int()` — convert to integer
- `float()` — convert to floating-point number
- `str()` — convert to string
- `bool()` — convert to boolean

````python
# str -> int -> float -> str
marks_text = "85"
marks_number = int(marks_text)
marks_float = float(marks_number)
marks_back_to_text = str(marks_float)

print(marks_number, type(marks_number))
print(marks_float, type(marks_float))
print(marks_back_to_text, type(marks_back_to_text))
````

````python
# bool() conversion rules: 0, 0.0, "", None, empty collections -> False. Everything else -> True
print(bool(0))
print(bool(1))
print(bool(""))
print(bool("Python"))
print(bool([]))
````

**Common mistake:** Trying `int("85.5")` directly raises a `ValueError`, because `"85.5"` is not a whole-number string. You must go through `float()` first: `int(float("85.5"))`. Try both below.

````python
# int("85.5")               # Uncomment to see the ValueError
fixed = int(float("85.5"))  # Correct approach
print(fixed)
````

### ML Connection
Datasets are frequently loaded as text (from CSV files, for example). Before any numeric computation or model training, that text must be converted into `int`/`float`. Pandas will automate a lot of this, but understanding the underlying conversion rules is essential when things go wrong.

# 13. Arithmetic Operators

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` | Addition | `5 + 2` | `7` |
| `-` | Subtraction | `5 - 2` | `3` |
| `*` | Multiplication | `5 * 2` | `10` |
| `/` | Division (always returns float) | `5 / 2` | `2.5` |

````python
a, b = 17, 5

print("a + b  =", a + b)
print("a - b  =", a - b)
print("a * b  =", a * b)
print("a / b  =", a / b)
print("a // b =", a // b)
print("a % b  =", a % b)
print("a ** b =", a ** b)
````

### ML Connection

<a id="comparison"></a>
# 14. Comparison Operators

Comparison operators compare two values and always return a `bool` (`True` or `False`).

| Operator | Meaning |
|---|---|
| `==` | equal to |
| `!=` | not equal to |
| `>` | greater than |
| `<` | less than |
| `>=` | greater than or equal to |
| `<=` | less than or equal to |

````python
score = 85

print(score == 85)
print(score != 90)
print(score > 90)
print(score >= 85)
````

**Common mistake:** Confusing `=` (assignment) with `==` (comparison). `if score = 85:` is a `SyntaxError` — you need `if score == 85:`.

<a id="logical"></a>
# 15. Logical Operators

Logical operators combine multiple boolean conditions.

| Operator | Meaning |
|---|---|
| `and` | True only if **both** conditions are True |
| `or` | True if **at least one** condition is True |
| `not` | Reverses a boolean value |

````python
attendance = 90     # percent
score = 78

is_eligible = attendance >= 75 and score >= 50
print("Eligible for certificate:", is_eligible)

print(not is_eligible)
print(attendance < 50 or score >= 50)
````

### ML Connection
Logical operators are the foundation of **filtering data** — for example, later in Pandas: "keep only rows where age > 18 and income > 50000".

<a id="assignment"></a>
# 16. Assignment Operators

Assignment operators combine an operation with assignment, letting you update a variable in place.

````python
total = 10
total += 5   # same as total = total + 5
print(total)

total -= 3   # same as total = total - 3
print(total)

total *= 2   # same as total = total * 2
print(total)

total /= 4   # same as total = total / 4
print(total)
````

### ML Connection
You'll see patterns like `total_loss += batch_loss` constantly when accumulating values across a loop — for example, summing up errors during model training.

<a id="string-slicing"></a>
# 18. String Indexing and Slicing

Strings are **sequences**, meaning each character has a position (an index), starting at `0`.

```
 P   y   t   h   o   n
 0   1   2   3   4   5
-6  -5  -4  -3  -2  -1
```

Negative indices count from the end: `-1` is the last character.

````python
word = "Python"

print(word[0])    # first character
print(word[5])    # last character (by position)
print(word[-1])   # last character (negative indexing)
print(word[-6])   # first character (negative indexing)
````

````python
# Slicing: word[start:stop] -- stop index is NOT included
word = "Python"

print(word[0:3])   # 'Pyt'
print(word[2:])    # from index 2 to the end -> 'thon'
print(word[:4])    # from the start up to (not including) index 4 -> 'Pyth'
print(word[::-1])  # step of -1 reverses the string -> 'nohtyP'
````

**Common mistake:** Trying `word[10]` on a 6-character string raises an `IndexError: string index out of range`. Always make sure your index is within bounds, especially when working with data of unknown length.

<a id="string-methods"></a>
# 19. Important String Methods

Strings come with many built-in methods. Here are the ones you'll use most often when cleaning text data.

````python
text = "  Machine Learning  "

print(text.upper())        # "  MACHINE LEARNING  "
print(text.lower())        # "  machine learning  "
print(text.strip())        # "Machine Learning" (removes leading/trailing whitespace)
print(text.replace("Machine", "Deep"))  # "  Deep Learning  "
print(text.strip().split())  # ['Machine', 'Learning'] -- splits into a list of words
````

````python
email = "student@example.com"

print(email.startswith("student"))
print(email.endswith(".com"))
print(email.find("@"))       # index of first occurrence, or -1 if not found
print("PYTHON".isupper())
print("python".islower())
````

### ML Connection
Real-world text data (like tweets, reviews, or CSV columns) is messy — extra spaces, inconsistent capitalization, unwanted symbols. String methods like `.strip()`, `.lower()`, and `.replace()` are your first tools for **data cleaning**, a huge part of any real ML project.

<a id="fstrings"></a>
# 20. f-strings (Formatted String Literals)

f-strings are the modern, preferred way to build strings that include variable values. Prefix the string with `f` and put variables inside `{}`.

````python
name = "Sara"
score = 92.456

print(f"Student {name} scored {score} marks.")
print(f"Rounded score: {score:.1f}")   # format to 1 decimal place
print(f"{name} passed: {score >= 50}")  # expressions work too!
````

**Output:**
```
Student Sara scored 92.456 marks.
Rounded score: 92.5
Sara passed: True
```

### ML Connection
When printing model results (accuracy, loss, predictions), you'll constantly use f-strings to build clean, readable reports, e.g. `f"Accuracy: {accuracy:.2%}"`.

<a id="list-slicing"></a>
# 22–24. List Indexing, Negative Indexing, and Slicing

Lists use the **same indexing and slicing rules as strings** — this consistency is one of Python's strengths.

````python
marks = [78, 85, 91, 67, 88]

print(marks[0])     # first element -> 78
print(marks[-1])    # last element -> 88
print(marks[1:4])   # elements at index 1,2,3 -> [85, 91, 67]
print(marks[:3])    # first three elements -> [78, 85, 91]
print(marks[::-1])  # reversed list
````

<a id="modifying-lists"></a>
# 25. Modifying Lists

Unlike strings, lists are **mutable** — you can change their contents after creation.

````python
marks = [78, 85, 91, 67, 88]

marks[0] = 80          # change a single element
print(marks)

marks.append(95)       # add to the end
print(marks)

marks.insert(1, 100)   # insert 100 at index 1
print(marks)

marks.remove(67)       # remove the first occurrence of 67
print(marks)

removed_value = marks.pop()   # removes and returns the last element
print(removed_value, marks)
````

**Common mistake:** Confusing `.remove(value)` (removes by *value*) with `.pop(index)` (removes by *position*). `marks.remove(2)` tries to remove the value `2`, not the element at index `2` — if `2` isn't in the list, this raises a `ValueError`.

<a id="list-methods"></a>
# 26. Important List Methods (Summary)

| Method | Purpose |
|---|---|
| `.append(x)` | add `x` to the end |
| `.insert(i, x)` | insert `x` at index `i` |
| `.remove(x)` | remove first occurrence of value `x` |
| `.pop(i)` | remove and return item at index `i` (default: last) |
| `.sort()` | sort the list in place |
| `.reverse()` | reverse the list in place |
| `.count(x)` | count occurrences of `x` |
| `.index(x)` | find index of first occurrence of `x` |

````python
marks = [78, 85, 91, 67, 88]

marks.sort()
print("Sorted:", marks)

marks.reverse()
print("Reversed:", marks)

print("Count of 88:", marks.count(88))
print("Index of 91:", marks.index(91))
````

<a id="nested-lists"></a>
# 27. Nested Lists

A **nested list** is a list that contains other lists. This is a very natural way to represent tabular data — a small preview of what Pandas DataFrames will do much more powerfully.

````python
# Each inner list = one student's marks in 3 subjects
student_marks = [
    [78, 85, 90],   # Ali
    [88, 92, 95],   # Sara
    [55, 62, 58],   # Ahmed
]

print(student_marks[0])       # Ali's marks
print(student_marks[0][1])    # Ali's 2nd subject mark -> 85
print(student_marks[1][-1])   # Sara's last subject mark -> 95
````

### ML Connection
This "list of lists" structure is conceptually identical to a small dataset table (rows = students, columns = subjects). NumPy 2D arrays and Pandas DataFrames are the specialized, high-performance versions of exactly this idea — you're already thinking the right way!

<a id="builtins"></a>
# 28. Built-in Functions Useful for ML

These five functions will appear constantly once you start working with real data.

````python
marks = [78, 85, 91, 67, 88]

print("Count:", len(marks))
print("Sum:", sum(marks))
print("Minimum:", min(marks))
print("Maximum:", max(marks))
print("Sorted:", sorted(marks))            # returns a NEW sorted list (original untouched)
print("Sorted descending:", sorted(marks, reverse=True))
print("Average:", sum(marks) / len(marks))  # combining sum() and len()
print("Original list unchanged:", marks)
````

**Common mistake:** `marks.sort()` sorts **in place** and returns `None`, while `sorted(marks)` returns a **new** sorted list and leaves the original untouched. Writing `marks = marks.sort()` is a classic bug — it sets `marks` to `None`!

### ML Connection
> "In Machine Learning, we frequently work with collections of values. Later, NumPy arrays and Pandas Series/DataFrames will provide much more powerful and efficient ways to work with such data — but the core ideas of summing, averaging, finding min/max, and sorting stay exactly the same.
