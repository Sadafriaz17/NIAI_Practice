## Task 2 — High Value, Long-Term Flag

For **each** customer, compute and print whether they are a **"high value, long-term"** customer, defined as:

```
tenure_months > 12   AND   monthly_charges > 70
```

Print a clear, labeled result for each customer (e.g. `"Customer 1 is high value & long-term: True"`).


````python
# Customer 1 check
customer1_flag = customer1_tenure > 12 and customer1_charges > 70
print("Customer 1 is high value & long-term:", customer1_flag)

# Customer 2 check
customer2_flag = customer2_tenure > 12 and customer2_charges > 70
print("Customer 2 is high value & long-term:", customer2_flag)

# Customer 3 check
customer3_flag = customer3_tenure > 12 and customer3_charges > 70
print("Customer 3 is high value & long-term:", customer3_flag)
````

## Task 3 — Bonus: Type Casting

Pick **one** customer's `monthly_charges` and cast it to `int`.

1. Print the original float value.
2. Print the casted int value.
3. In a comment, explain what value was lost and why (hint: `int()` truncates, it does not round).


````python
# Cast Customer 1's monthly charges from float to int
original_value = customer1_charges
casted_value = int(customer1_charges)

print("Original float value:", original_value)
print("Casted int value:", casted_value)

# The .95 part was lost. int() truncates (cuts off) everything after the
# decimal point. It does not round, so 89.95 becomes 89, not 90.
````
