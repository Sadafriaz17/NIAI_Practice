```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def average(self):
        return sum(self.marks) / len(self.marks)

    def grade(self):
        avg = self.average()

        if avg >= 90:
            return "A grade"
        elif avg >= 80:
            return "B grade"
        elif avg >= 70:
            return "C grade"
        else:
            return "F grade"


Student1 = Student("Ali", [90, 80, 70])

print(Student1.name, Student1.marks)
print("Average marks:", Student1.average())
print("Grade:", Student1.grade())


```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def average(self):
        return sum(self.marks) / len(self.marks)


students = [
    Student("Ali", [78, 85, 90]),
    Student("Sara", [88, 92, 95]),
    Student("Ahmed", [55, 62, 58]),
]

for s in students:
    print(f"{s.name}: {s.average():.2f}")
```

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance

    def deposit(self, amount):
        if amount > 0:
            print(f"Depositing {amount} to {self.owner}'s account.")
            self._balance += amount

    def get_balance(self):
        return self._balance


account = BankAccount("Ali", 1000)

account.deposit(500)
print(account.get_balance())

account._balance = 999999
print(account.get_balance())   
  
```
