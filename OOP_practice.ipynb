```python
class Student:
    school_name="AIML academy"

    def __init__(self, name, age):
        self.name=name
        self.age=age

Student1=Student("Sadaf", 22)
Student2=Student("Ifra", 20)

print(Student1.name, Student1.age, Student1.school_name)
print(Student2.name, Student2.age, Student2.school_name)

Student1.school_name = "Data Science Institute"
print(Student1.school_name)  
print(Student2.school_name)
```
