# 2. Conditional Statements

Conditionals let a program make decisions and run different code depending on whether a condition is `True` or `False`.

- `if` — run a block if a condition is True
- `elif` ("else if") — check another condition if the previous one was False
- `else` — run this block if none of the above conditions were True

Remember: each block must be **indented** consistently (see Part 1).

````python
score = 72

if score >= 90:
    grade = "A"
elif score >= 75:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "F"

print(f"Score: {score}, Grade: {grade}")
````

**Common mistake:** Using `if` for every condition instead of `elif`. If you write separate `if` statements instead of `elif`, Python checks *all* of them (even after one is already True) — which can lead to bugs or wasted computation.

<a id="nested-conditions"></a>
# 3. Nested Conditions

You can place an `if` statement inside another `if` statement. Use this when a decision depends on more than one level of logic. Be careful — too much nesting hurts readability; often `and`/`or` (from Part 1) is a cleaner alternative.

````python
age = 20
has_id_card = True

if age >= 18:
    if has_id_card:
        print("Entry allowed.")
    else:
        print("Entry denied: ID card required.")
else:
    print("Entry denied: must be 18 or older.")

# The same logic, often preferred, using 'and':
if age >= 18 and has_id_card:
    print("Entry allowed. (using 'and')")
````

### ML Connection
Filtering and rule-based logic — like nested conditions — is exactly what you'll later express as boolean masks in Pandas (e.g. `df[(df["age"] >= 18) & (df["has_id"] == True)]`).

<a id="for-loops"></a>
# 4. `for` Loops and `range()`

A `for` loop repeats a block of code once for each item in a sequence (like a list, string, or range of numbers).

`range(start, stop, step)` generates a sequence of numbers. `stop` is **not included**, just like slicing.

````python
for i in range(5):        # 0,1,2,3,4
    print(i)
````

````python
for i in range(2, 10, 2):  # start=2, stop=10 (excluded), step=2
    print(i)
````

````python
marks = [78, 85, 91, 67, 88]

total = 0
for mark in marks:
    total += mark

print("Total:", total)
print("Average:", total / len(marks))
````

**Common mistake:** Assuming `range(5)` includes `5` — it does not. `range(5)` produces `0, 1, 2, 3, 4`.

<a id="while-loops"></a>
# 5. `while` Loops

A `while` loop repeats a block of code **as long as** a condition remains True. Unlike `for` loops (which iterate a known number of times), `while` loops are useful when you don't know in advance how many iterations you need.

````python
count = 0

while count < 5:
    print("Count is", count)
    count += 1   # IMPORTANT: without this, the loop would run forever!

print("Loop finished.")
````

**Common mistake:** Forgetting to update the loop variable (`count += 1` above) creates an **infinite loop**, which will freeze your notebook. If this ever happens, use the "Interrupt Kernel" / "Stop" button.

<a id="break-continue"></a>
# 6. `break` and `continue`

- `break` — immediately exits the loop entirely.
- `continue` — skips the rest of the current iteration and moves to the next one.

````python
# break example: stop as soon as we find a failing score
scores = [88, 92, 45, 76, 90]

for score in scores:
    if score < 50:
        print(f"Found a failing score ({score}). Stopping check.")
        break
    print(f"{score} is OK")
````

````python
# continue example: skip negative numbers, only sum positive ones
values = [10, -5, 20, -1, 30]
total = 0

for v in values:
    if v < 0:
        continue   # skip the rest of this iteration
    total += v

print("Sum of positive values:", total)
````

# 8. Dictionaries

A **dictionary** (`dict`) stores data as **key-value pairs**. Unlike a list (which uses numeric positions), a dictionary uses meaningful keys to access values — perfect for representing a single structured record.

````python
student = {
    "name": "Ali",
    "age": 22,
    "marks": [78, 85, 90]
}

print(student)
print(type(student))
print(student["name"])
print(student["marks"])
````

````python
# Accessing a missing key with [] raises a KeyError -- .get() is safer
print(student.get("grade"))            # returns None instead of crashing
print(student.get("grade", "N/A"))     # returns a default value instead

# Modifying / adding
student["age"] = 23          # update existing key
student["grade"] = "A"       # add a new key
print(student)
````

## Dictionary Keys, Values, and Methods

````python
print(student.keys())     # all keys
print(student.values())   # all values
print(student.items())    # all (key, value) pairs

for key, value in student.items():
    print(f"{key}: {value}")
````

**Common mistake:** Using `student["grade"]` when `"grade"` doesn't exist raises a `KeyError`. Prefer `.get()` when a key might be missing — especially with real-world, messy data.

<a id="nested-dicts"></a>
# 9. Nested Dictionaries

Dictionaries can contain other dictionaries (or lists) as values. This is exactly the structure of real-world JSON data — configuration files, API responses, and datasets.

````python
students = {
    "Ali": {"age": 22, "marks": [78, 85, 90]},
    "Sara": {"age": 21, "marks": [88, 92, 95]},
}

print(students["Ali"]["marks"])          # Ali's marks
print(students["Sara"]["age"])           # Sara's age

for name, info in students.items():
    avg = sum(info["marks"]) / len(info["marks"])
    student_age = info["age"]
    print(f"{name}: age {student_age}, average {avg:.2f}")
````

### ML Connection
> Dictionaries map directly to structured records, configuration files, and JSON data — you'll use this exact shape when configuring ML experiments (e.g. `{"model": "logistic_regression", "learning_rate": 0.01, "epochs": 100}`) and when loading real-world datasets from JSON APIs.

# 10. Tuples

A **tuple** is like a list, but **immutable** — once created, it cannot be changed. Tuples are written with parentheses `()`. Use them for fixed collections of values that shouldn't be modified, like coordinates or a fixed-size record.

````python
point = (10, 20)
print(point[0], point[1])

# Tuples are immutable:
# point[0] = 99   # Uncomment to see the TypeError

# Common use: returning multiple values from a function (covered later),
# or unpacking:
x, y = point
print(f"x={x}, y={y}")
````

### ML Connection
Tuples are often used for things that shouldn't change during a program, such as an image's `(width, height)` or a dataset's fixed `(rows, columns)` shape — a concept you'll see constantly once you reach NumPy array shapes.

<a id="sets"></a>
# 11. Sets

A **set** is an unordered collection of **unique** values. Sets automatically remove duplicates and are useful for membership checks and comparing collections.

````python
grades = ["A", "B", "A", "C", "B", "A"]

unique_grades = set(grades)
print(unique_grades)          # duplicates removed
print(len(unique_grades))     # number of unique grades

print("A" in unique_grades)   # fast membership check
````

````python
# Set operations
class_a_students = {"Ali", "Sara", "Ahmed"}
class_b_students = {"Sara", "Ayesha", "Bilal"}

print("In both classes:", class_a_students & class_b_students)   # intersection
print("In either class:", class_a_students | class_b_students)   # union
print("Only in class A:", class_a_students - class_b_students)   # difference
````

### ML Connection
Sets are handy for tasks like finding unique categories in a feature column (e.g. unique city names) or comparing which labels appear in a training set versus a test set.

<a id="function-details"></a>
# 13. Parameters, Arguments, `return`, and Default Parameters

- **Parameter**: a variable listed in the function's definition.
- **Argument**: the actual value you pass in when calling the function.
- `return`: sends a value back to the caller (without it, a function returns `None`).
- **Default parameter**: a parameter with a pre-set value, used if the caller doesn't provide one.

````python
def calculate_average(marks):        # 'marks' is a parameter
    return sum(marks) / len(marks)

result = calculate_average([78, 85, 90])   # [78, 85, 90] is the argument
print(result)
````

````python
def describe_student(name, average, passing_mark=50):
    # 'passing_mark' has a default value -- caller can override it
    status = "Passed" if average >= passing_mark else "Failed"
    return f"{name}: average={average:.1f}, status={status}"

print(describe_student("Ali", 84.3))                 # uses default passing_mark=50
print(describe_student("Ahmed", 45.0))
print(describe_student("Bilal", 55.0, passing_mark=60))  # overrides the default
````

**Common mistake:** Forgetting `return` and expecting the function to "give back" a value. Without `return`, a function performs its work (e.g. printing) but the function call itself evaluates to `None`.

### ML Connection
> Functions map to reusable preprocessing/calculation logic — for example, a `normalize(value, min_val, max_val)` function you'll write and reuse across an entire ML project.

<a id="scope"></a>
# 14. Local vs Global Variables (Basic Understanding)

- A **local variable** is created inside a function and only exists while that function runs.
- A **global variable** is created outside any function and can be read from anywhere in the notebook.

You generally shouldn't modify a global variable from inside a function unless you explicitly use the `global` keyword (which we recommend avoiding as a beginner — pass values in and return values out instead).

````python
threshold = 50   # global variable

def is_passing(mark):
    # 'mark' is local to this function; 'threshold' is read from the global scope
    return mark >= threshold

print(is_passing(65))
print(is_passing(40))

# 'mark' does not exist outside the function:
# print(mark)   # Uncomment to see the NameError
````

**Common mistake:** Expecting a variable created inside a function to be usable outside it. Once the function finishes, its local variables are gone.

<a id="enumerate-zip"></a>
# 15. `enumerate()` and `zip()`

- `enumerate(sequence)` — loop over a sequence while also getting the index of each item.
- `zip(seq1, seq2, ...)` — loop over multiple sequences in parallel, pairing up corresponding items.

````python
subjects = ["Math", "Science", "English"]

for index, subject in enumerate(subjects):
    print(index, subject)

print()

for index, subject in enumerate(subjects, start=1):  # start counting from 1
    print(index, subject)
````

````python
names = ["Ali", "Sara", "Ahmed"]
scores = [84, 92, 58]

for name, score in zip(names, scores):
    print(f"{name}: {score}")

# Building a dictionary from two lists using zip()
name_to_score = dict(zip(names, scores))
print(name_to_score)
````

### ML Connection
`zip()` is exactly the idea behind pairing up **features and labels**, or predicted values and actual values, when checking model performance — for example: `for predicted, actual in zip(predictions, true_values):`

# 16. List and Dictionary Comprehensions

**Comprehensions** are a compact, Pythonic way to build a new list or dictionary from an existing sequence, optionally filtering or transforming elements. They replace many simple `for` loops with a single readable line.

Syntax:
```
[expression for item in sequence if condition]
{key_expr: value_expr for item in sequence if condition}
```

````python
marks = [78, 85, 91, 67, 88]

# Traditional loop
squared = []
for m in marks:
    squared.append(m ** 0.5)   # square root, just as an example transform

# Equivalent list comprehension
squared_comp = [m ** 0.5 for m in marks]

print(squared)
print(squared_comp)
````

````python
marks = [78, 85, 91, 67, 88]

# Filtering with a condition
passing_marks = [m for m in marks if m >= 75]
print("Passing marks:", passing_marks)

# Transforming AND filtering
grades = ["Pass" if m >= 75 else "Fail" for m in marks]
print("Grades:", grades)
````

````python
# Dictionary comprehension -- basic introduction
names = ["Ali", "Sara", "Ahmed"]
scores = [84, 92, 58]

score_dict = {name: score for name, score in zip(names, scores)}
print(score_dict)

# Filter while building a dictionary
passing_dict = {name: score for name, score in score_dict.items() if score >= 60}
print(passing_dict)
````

**Common mistake:** Overusing comprehensions until code becomes unreadable. If a comprehension needs more than one condition or transformation to stay clear, a regular `for` loop is often the better, more maintainable choice.

# 19. Exception Handling

An **exception** is an error that occurs while a program is running. If unhandled, it stops the program. Python lets you **catch** exceptions gracefully using `try`/`except`/`else`/`finally`.

- `try`: code that might raise an error
- `except`: runs if an error occurs
- `else`: runs only if **no** error occurred
- `finally`: always runs, error or not (useful for cleanup)

````python
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Error: cannot divide by zero.")
        return None
    else:
        print("Division succeeded.")
        return result
    finally:
        print("safe_divide() finished running.")

print(safe_divide(10, 2))
print()
print(safe_divide(10, 0))
````

````python
# Catching multiple, specific exception types is good practice
def to_number(text):
    try:
        return int(text)
    except ValueError:
        print(f"'{text}' is not a valid whole number.")
        return None

print(to_number("42"))
print(to_number("forty-two"))
````

**Common mistake:** Writing a bare `except:` that catches *every* possible error. This hides real bugs. Always catch the **specific** exception type you expect (`ValueError`, `ZeroDivisionError`, `KeyError`, etc.).

### ML Connection
Real datasets are messy: missing values, malformed rows, unexpected types. Exception handling lets your data-loading and preprocessing code skip or report bad records instead of crashing your entire pipeline.

# 20. File Handling

Python can read and write plain text files using the built-in `open()` function. The `with` statement automatically closes the file for you, even if an error occurs — always prefer it over manually calling `.close()`.

````python
# Writing to a text file
with open("sample_data/notes.txt", "w") as file:
    file.write("Machine Learning Notes\n")
    file.write("Python is the foundation.\n")
    file.write("NumPy and Pandas come next.\n")

print("File written successfully.")
````

````python
# Reading line by line (memory-efficient for large files)
with open("sample_data/notes.txt", "r") as file:
    for line in file:
        print("Line:", line.strip())   # .strip() removes the trailing newline character
````

````python
# Appending to an existing file instead of overwriting it
with open("sample_data/notes.txt", "a") as file:
    file.write("Appended: Practice consistently!\n")

with open("sample_data/notes.txt", "r") as file:
    print(file.read())
````

# 21. Serialization and Deserialization

These two words sound intimidating, but the idea is simple:

- **Serialization**: converting a Python object (like a dictionary or list) into a format that can be **stored** (in a file) or **transmitted** (over a network).
- **Deserialization**: converting that stored/transmitted format **back** into a Python object.

```
Python Object
     ↓   (serialization)
Stored Data  (e.g. a JSON string, or bytes on disk)
     ↓   (deserialization)
Python Object
```

We'll use two standard-library tools:
- **`json`** — for interoperable, human-readable data exchange.
- **`pickle`** — for Python-specific, general-purpose object storage.

### Why does this matter for ML?
- Saving configuration (e.g. hyperparameters) so an experiment can be reproduced.
- Saving preprocessing information (e.g. the min/max values used to scale a feature).
- Storing experiment metadata (dates, results, notes).
- Saving Python objects between sessions, so you don't have to recompute everything.
- Persisting trained models conceptually (in real projects, this is often done with tools built specifically for that, but the underlying idea — serialize now, deserialize later — is the same).
- Sending/storing structured data, such as results shared between team members or systems.

<a id="json-section"></a>
# 22. JSON Serialization

**JSON** (JavaScript Object Notation) is a lightweight, text-based, language-independent data format. It looks a lot like Python dictionaries and lists, which makes it very natural to work with.

The `json` module (standard library) gives us four main functions:

| Function | Purpose |
|---|---|
| `json.dumps(obj)` | Python object → JSON **string** |
| `json.loads(text)` | JSON string → Python object |
| `json.dump(obj, file)` | Python object → JSON written directly to a **file** |
| `json.load(file)` | JSON read directly from a file → Python object |

````python
import json

student = {
    "name": "Ali",
    "age": 22,
    "marks": [80, 85, 90]
}

serialized = json.dumps(student)          # Python dict -> JSON string
print(serialized)
print(type(serialized))
````

````python
deserialized = json.loads(serialized)     # JSON string -> Python dict
print(deserialized)
print(type(deserialized))

print(deserialized["marks"][0])   # it behaves like a normal Python dict again
````

````python
# Pretty-printing JSON for readability, with indentation
pretty = json.dumps(student, indent=4)
print(pretty)
````

````python
# Writing JSON directly to a file
with open("sample_data/student.json", "w") as file:
    json.dump(student, file, indent=4)

print("student.json written.")

# Reading JSON directly from a file
with open("sample_data/student.json", "r") as file:
    loaded_student = json.load(file)

print(loaded_student)
print(type(loaded_student))
````

**Common mistake:** Confusing `dumps`/`loads` (string versions, note the trailing **s** for "string") with `dump`/`load` (file versions, no trailing "s"). Also, JSON has no `tuple` type — tuples are automatically converted to JSON arrays (lists) during serialization.

<a id="pickle-section"></a>
# 23. Pickle Serialization

**Pickle** is Python's own, built-in serialization format. Unlike JSON, pickle:
- Can serialize almost **any** Python object (not just dicts, lists, numbers, and strings).
- Produces **binary** data, not human-readable text.
- Is **Python-specific** — a pickle file generally can't be read by other languages.

Because pickle writes binary data, always open files in `"wb"` (write-binary) and `"rb"` (read-binary) mode.

````python
import pickle

data = {
    "name": "Ali",
    "scores": [80, 85, 90]
}

with open("sample_data/data.pkl", "wb") as file:
    pickle.dump(data, file)

print("data.pkl written.")
````

````python
with open("sample_data/data.pkl", "rb") as file:
    loaded_data = pickle.load(file)

print(loaded_data)
print(type(loaded_data))
````

## JSON vs Pickle — When to Use Which

| | JSON | Pickle |
|---|---|---|
| Human-readable? | Yes | No (binary) |
| Cross-language? | Yes | No (Python only) |
| Can store almost any Python object? | No (dict, list, str, int, float, bool, None only) | Yes |
| Typical ML use | Configs, metadata, results shared across tools | Saving arbitrary Python objects for later reuse *within* Python |

<a id="pickle-warning"></a>
# 24. ⚠️ Important Security Warning: Pickle

> **Never unpickle data from an untrusted or unknown source.**

Here's why this matters: unlike JSON (which is just structured text), **pickle can execute arbitrary code during deserialization**. A malicious `.pkl` file can be crafted to run harmful commands on your computer the moment you call `pickle.load()` — even before you look at the "data" it claims to contain.

**Safe practice:**
- Only unpickle files that *you* created, or that come from a fully trusted, verified source.
- If you need to exchange data with others (or across systems/languages), prefer **JSON** instead.
- Never load a `.pkl` file downloaded from an unknown website or received from an untrusted sender.

# Final Project: Student Performance Analyzer

This project ties together everything from **both** notebooks: variables, data types, lists, dictionaries, conditions, loops, functions, comprehensions, exception handling, file handling, and JSON serialization.

**Goal:** Given a list of students and their marks, we will:
1. Calculate each student's average.
2. Assign a simple grade.
3. Identify students above a chosen threshold.
4. Calculate the class average.
5. Display a summary.
6. Save the results to a JSON file.
7. Load the JSON file again.
8. Display the restored data.

This is exactly the kind of small, structured, end-to-end script you'll write dozens of times while preparing datasets for real ML models.

````python
students = [
    {"name": "Ali", "marks": [78, 85, 90]},
    {"name": "Sara", "marks": [88, 92, 95]},
    {"name": "Ahmed", "marks": [55, 62, 58]},
    {"name": "Ayesha", "marks": [90, 87, 93]},
]

print(f"Loaded {len(students)} student records.")
````

````python
def calculate_average(marks):
    """Return the average of a list of marks. Returns 0 for an empty list."""
    if not marks:
        return 0
    return sum(marks) / len(marks)


def assign_grade(average):
    """Assign a simple letter grade based on an average score."""
    if average >= 90:
        return "A"
    elif average >= 75:
        return "B"
    elif average >= 60:
        return "C"
    else:
        return "F"


def analyze_students(students, threshold=80):
    """
    Process a list of student records and return an enriched list
    that includes each student's average and grade.
    """
    analyzed = []
    for student in students:
        try:
            average = calculate_average(student["marks"])
        except (KeyError, TypeError) as error:
            print(f"Skipping malformed record {student}: {error}")
            continue

        grade = assign_grade(average)
        analyzed.append({
            "name": student["name"],
            "marks": student["marks"],
            "average": round(average, 2),
            "grade": grade,
            "above_threshold": average >= threshold,
        })
    return analyzed


analyzed_students = analyze_students(students, threshold=80)
for record in analyzed_students:
    print(record)
````

````python
# Step 3: Identify students above the threshold (using a list comprehension)
top_students = [s["name"] for s in analyzed_students if s["above_threshold"]]
print("Students above threshold:", top_students)

# Step 4: Calculate the class average
class_average = sum(s["average"] for s in analyzed_students) / len(analyzed_students)
print(f"Class average: {class_average:.2f}")
````

````python
# Step 5: Display a clean summary report
print("=" * 40)
print("STUDENT PERFORMANCE SUMMARY")
print("=" * 40)

for s in analyzed_students:
    status = "Above threshold" if s["above_threshold"] else "Below threshold"
    name = s["name"]
    avg = s["average"]
    grade = s["grade"]
    print(f"{name:<8} | Avg: {avg:>6} | Grade: {grade} | {status}")

print("-" * 40)
print(f"Class Average: {class_average:.2f}")
top_performers_text = ", ".join(top_students)
print(f"Top performers: {top_performers_text}")
print("=" * 40)
````

````python
import json

# Step 6: Save the results to a JSON file
report = {
    "students": analyzed_students,
    "class_average": round(class_average, 2),
    "top_performers": top_students,
}

with open("sample_data/performance_report.json", "w") as file:
    json.dump(report, file, indent=4)

print("Report saved to sample_data/performance_report.json")
````

````python
# Step 7: Load the JSON file back
with open("sample_data/performance_report.json", "r") as file:
    restored_report = json.load(file)

# Step 8: Display the restored data
print("Restored class average:", restored_report["class_average"])
print("Restored top performers:", restored_report["top_performers"])
print()
for s in restored_report["students"]:
    print(s)
````
